```python
from tensorflow.keras import layers, models
############################ Encoder ################################

inputs = layers.Input(shape=(256, 256, 3))
conv1 = layers.Conv2D(64,3,activation='relu',padding='same')(inputs)
conv1 = layers.Conv2D(64,3,activation='relu',padding='same')(conv1)
pool1 = layers.MaxPool2D(pool_size=(2,2))(conv1)

conv2 = layers.Conv2D(128,3,activation='relu', padding='same')(pool1)
conv2 = layers.Conv2D(128,3,activation='relu', padding='same')(conv2)
pool2 = layers.MaxPool2D(pool_size=(2,2))(conv2)

conv3 = layers.Conv2D(256,3,activation='relu',padding='same')(pool2)
conv3 = layers.Conv2D(256,3, activation='relu',padding='same')(conv3)
pool3 = layers.MaxPool2D(pool_size=(2,2))(conv3)

conv4 = layers.Conv2D(512,3,activation='relu',padding='same')(pool3) 
conv4 = layers.Conv2D(512,3,activation='relu',padding='same')(conv4)
pool4 = layers.MaxPool2D(pool_size=(2,2))(conv4)


########################## Deep Layer ######################
conv5 = layers.Conv2D(1024,3,activation='relu',padding='same')(pool4)
conv5 = layers.Conv2D(1024,3, activation='relu',padding='same')(conv5)

########################### Decoder ########################

up1 = layers.Conv2DTranspose(512,2,strides=2,padding='same')(conv5)
merge1 = layers.Concatenate()([conv4,up1])
conv6 = layers.Conv2D(512,3,activation='relu',padding='same')(merge1)
conv6 = layers.Conv2D(512,3,activation='relu',padding='same')(conv6)


up2 = layers.Conv2DTranspose(256,2,strides=2,padding='same')(conv6)
merge2 = layers.Concatenate()([conv3,up2])
conv7 = layers.Conv2D(256,3,activation='relu',padding='same')(merge2)
conv7 = layers.Conv2D(256,3,activation='relu',padding='same')(conv7)


up3 = layers.Conv2DTranspose(128,2,strides=2,padding='same')(conv7)
merge3 = layers.Concatenate()([conv2,up3])
conv8 = layers.Conv2D(128,3,activation='relu',padding='same')(merge3)
conv8 = layers.Conv2D(128,3,activation='relu',padding='same')(conv8)


up4 = layers.Conv2DTranspose(64,2,strides=2,padding='same')(conv8)
merge4 = layers.Concatenate()([conv1,up4])
conv9 = layers.Conv2D(64,3,activation='relu',padding='same')(merge4)
conv9 = layers.Conv2D(64,3,activation='relu', padding='same')(conv9)

outputs = layers.Conv2D(7,1, activation='softmax')(conv9)
model = models.Model(inputs=inputs, outputs=outputs)

model.summary()










```

```python
def load_data(image_name, mask_name):
    
    img_path = tf.strings.join([image_path, "/", image_name])
    mask_path_full = tf.strings.join([mask_path, "/", mask_name])
    img_file = tf.io.read_file(img_path)
    png = tf.image.decode_png(img_file,channels=3)
    msk_file = tf.io.read_file(mask_path_full)
    png_mask = tf.image.decode_png(msk_file,channels=1)
    return png,png_mask
```

```python
def remap_dataset(image,mask):
    mask = tf.cast(mask , tf.int32)
    remapped_mask = tf.where(mask == 0, 255, mask-1)
    return image,remapped_mask
```

```python
def augmentation(image, mask):
    combine = tf.concat([image,tf.cast(mask,image.dtype)],axis=-1)
    combine = tf.image.random_flip_left_right(combine)
    combine = tf.image.random_flip_up_down(combine)
    k = tf.random.uniform(shape=[],minval=0,maxval = 4, dtype=tf.int32)
    combined = tf.image.rot90(combine, k)
    image = combine[:, :, :3]
    mask = combine[:, :, 3:]
    return image, tf.cast(mask,tf.uint8)
    
```

```python
def random_image(image,mask):
    combined = tf.concat([image, tf.cast(mask, image.dtype)], axis=-1)
    cropped = tf.image.random_crop(combined,size=[256,256,4])
    image_Crop = cropped[:,:,:3]
    mask_Crop = cropped[:,:,3:]
    return image_Crop, tf.cast(mask_Crop,mask.dtype)
    
```

```python
def fixed_crop(image, mask):

    top = 0
    left = 0

    image_crop = image[top:top+256, left:left+256, :]
    mask_crop = mask[top:top+256, left:left+256, :]

    return image_crop, mask_crop
```

```python
def masked_sparse_categorical_crossentropy(y_true, y_pred):

    # Remove the last dimension of the mask
    y_true = tf.squeeze(y_true, axis=-1)

    # Identify valid pixels (ignore label 255)
    valid_mask = tf.not_equal(y_true, 255)

    # Replace ignored labels with a safe class ID
    safe_labels = tf.where(valid_mask, y_true, 0)

    # Calculate cross-entropy for each pixel
    pixel_loss = tf.keras.losses.sparse_categorical_crossentropy(
        safe_labels,
        y_pred,
        from_logits=False
    )

    # Convert valid mask to float
    valid_mask = tf.cast(valid_mask, tf.float32)

    # Ignore losses belonging to no-data pixels
    pixel_loss = pixel_loss * valid_mask

    # Calculate average loss over valid pixels only
    total_loss = tf.reduce_sum(pixel_loss)
    valid_pixels = tf.reduce_sum(valid_mask)

    return total_loss / (valid_pixels + tf.keras.backend.epsilon())
```

```python
class MaskedMeanIoU(tf.keras.metrics.Metric):

    def __init__(self, num_classes=7, name="mean_iou", **kwargs):
        super().__init__(name=name, **kwargs)

        self.num_classes = num_classes

        self.confusion_matrix = self.add_weight(
            name="confusion_matrix",
            shape=(num_classes, num_classes),
            initializer="zeros"
        )

    def update_state(self, y_true, y_pred, sample_weight=None):

        y_true = tf.squeeze(y_true, axis=-1)

        y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

        y_true = tf.cast(y_true, tf.int32)

        valid_mask = tf.not_equal(y_true, 255)

        y_true = tf.boolean_mask(y_true, valid_mask)
        y_pred = tf.boolean_mask(y_pred, valid_mask)

        batch_confusion = tf.math.confusion_matrix(
            y_true,
            y_pred,
            num_classes=self.num_classes,
            dtype=tf.float32
        )

        self.confusion_matrix.assign_add(batch_confusion)

    def result(self):

        intersection = tf.linalg.diag_part(self.confusion_matrix)

        ground_truth = tf.reduce_sum(self.confusion_matrix, axis=1)
        predictions = tf.reduce_sum(self.confusion_matrix, axis=0)

        union = ground_truth + predictions - intersection

        iou = tf.math.divide_no_nan(intersection, union)

        return tf.reduce_mean(iou)

    def reset_state(self):
        self.confusion_matrix.assign(
            tf.zeros_like(self.confusion_matrix)
        )
```

```python
###Hold
def masked_multiclass_dice_loss(y_true, y_pred, smooth=1e-6):

    # Remove the last dimension from ground-truth masks
    y_true = tf.squeeze(y_true, axis=-1)
    y_true = tf.cast(y_true, tf.int32)

    # Ignore pixels labelled 255
    valid_mask = tf.not_equal(y_true, 255)

    # Replace ignored labels with 0 temporarily
    safe_labels = tf.where(valid_mask, y_true, 0)

    # Convert ground truth into one-hot format
    y_true_onehot = tf.one_hot(safe_labels, depth=7)

    # Convert valid mask to float and add class dimension
    valid_mask = tf.cast(valid_mask, tf.float32)
    valid_mask = tf.expand_dims(valid_mask, axis=-1)

    # Exclude ignored pixels from both tensors
    y_true_onehot = y_true_onehot * valid_mask
    y_pred = y_pred * valid_mask

    # Calculate intersection and denominator per class
    axes = [0, 1, 2]

    intersection = tf.reduce_sum(
        y_true_onehot * y_pred, axis=axes
    )

    denominator = tf.reduce_sum(
        y_true_onehot + y_pred, axis=axes
    )

    # Dice score for each class
    dice_score = (
        2.0 * intersection + smooth
    ) / (
        denominator + smooth
    )

    # Mean Dice across all 7 classes
    mean_dice = tf.reduce_mean(dice_score)

    return 1.0 - mean_dice
```

```python

def combined_loss(y_true, y_pred):

    ce_loss = weighted_masked_sparse_categorical_crossentropy(
        y_true, y_pred
    )

    dice_loss = masked_multiclass_dice_loss(
        y_true, y_pred
    )

    return ce_loss + dice_loss
```

```python
class_frequencies = class_counts / np.sum(class_counts)

class_weights = np.sqrt(
    np.max(class_frequencies) / class_frequencies
)

print("Class weights:", class_weights)
```

```python
def weighted_masked_sparse_categorical_crossentropy(y_true, y_pred):

    # Remove the last dimension of the mask
    y_true = tf.squeeze(y_true, axis=-1)

    # Identify valid pixels (ignore label 255)
    valid_mask = tf.not_equal(y_true, 255)

    # Replace ignored labels with a safe class ID
    safe_labels = tf.where(valid_mask, y_true, 0)
    safe_labels = tf.cast(safe_labels, tf.int32)

    # Calculate cross-entropy for each pixel
    pixel_loss = tf.keras.losses.sparse_categorical_crossentropy(
        safe_labels,
        y_pred,
        from_logits=False
    )

    # Define class weights
    class_weights_tensor = tf.constant(class_weights, dtype=tf.float32)

    # Assign the appropriate weight to each pixel
    pixel_weights = tf.gather(class_weights_tensor, safe_labels)

    # Convert valid mask to float
    valid_mask = tf.cast(valid_mask, tf.float32)

    # Apply class weights and ignore invalid pixels
    weighted_pixel_loss = pixel_loss * pixel_weights * valid_mask

    # Calculate weighted average loss over valid pixels
    total_loss = tf.reduce_sum(weighted_pixel_loss)
    total_weight = tf.reduce_sum(pixel_weights * valid_mask)

    return total_loss / (total_weight + tf.keras.backend.epsilon())
```
